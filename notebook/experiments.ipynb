```python
# vectorstore
from langchain.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
doc_vector=embedding_model.embed_documents(docs[0].page_content)
```

```python
vectorstore=FAISS.from_documents(docs, embedding_model)
```

```python
# RETREIVAL PROPCESS
relevant_doc=vectorstore.similarity_search("llama2 finetuning benchmark experiments.")
```

```python
# retriever=vectorstore.as_retriever() so that we can use it in chaining concept
retriever = vectorstore.as_retriever(search_kwargs={"k": 5})
retriever.invoke("explain about decoider.")
```

```python
from langchain.prompts import PromptTemplate
```

```python
prompt_template = """
        Answer the question based on the context provided below. 
        If the context does not contain sufficient information, respond with: 
        "I do not have enough information about this."

        Context: {context}

        Question: {question}

        Answer:"""
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variables=["context", "question"]
)
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
rag_chain.invoke("tell  me about decoder?")
```

```python
rag_chain.invoke("can you tell me Scaling trends for the reward model?")
```
