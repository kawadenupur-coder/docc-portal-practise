```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
from dotenv import load_dotenv
import os

# Load .env file
load_dotenv()

# Verify it loaded
api_key = os.getenv("GOOGLE_API_KEY")
print(f"API Key loaded: {api_key}")
print(f"API Key length: {len(api_key) if api_key else 0}")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# NOW create the model (after loading the API key)
embedding_model = GoogleGenerativeAIEmbeddings(
    model="models/embedding-001"
)

print("Embedding model created successfully!")
```

```python
doc_vector=embedding_model.embed_query("What is the capital of India?")
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

# Free, no API key needed, no rate limits!
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Test it
doc_vector = embedding_model.embed_query("What is the capital of India?")
print(f"✓ Success! Embedding dimension: {len(doc_vector)}")
print(f"✓ First 5 values: {doc_vector[:5]}")
```
